```python
import pygame
import sys

# Initialize pygame
pygame.init()

# Window size
SIZE = (1000, 1000)

# Create display window
display = pygame.display.set_mode(SIZE)

# Set window title
pygame.display.set_caption("Custom Cursor")

# Load cursor image
cursor = pygame.image.load("cursor.jpeg").convert_alpha()

# Resize cursor image
cursor = pygame.transform.scale(cursor, (50, 50))

# Hide default mouse cursor
pygame.mouse.set_visible(False)

# Create clock object for FPS control
clock = pygame.time.Clock()

# Main loop control variable
running = True

while running:

    # Fill screen with white color
    display.fill((255, 255, 255))

    # Check events
    for event in pygame.event.get():

        # Quit window
        if event.type == pygame.QUIT:
            running = False

        # Press ESC key to quit
        if event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE:
            running = False

    # Get current mouse position
    pos_x, pos_y = pygame.mouse.get_pos()

    # Center cursor image on mouse position
    pos_x -= cursor.get_width() // 2
    pos_y -= cursor.get_height() // 2

    # Draw custom cursor image
    display.blit(cursor, (pos_x, pos_y))

    # Update display
    pygame.display.flip()

    # Limit FPS to 60
    clock.tick(60)

# Quit pygame
pygame.quit()

# Exit program
#sys.exit()
```
