```python
# ==========================================
# Simple Game Menu with Player Information
# ==========================================

import tkinter as tk


# ==========================================
# Function: Open Game Screen
# ==========================================
def open_game_screen(player1_name, player2_name):
    """
    Opens the game screen after player names
    have been entered.
    """

    # Create a new window
    game_window = tk.Toplevel()

    # Configure window
    game_window.title("Game Screen")
    game_window.geometry("500x400")
    game_window.configure(bg="lightgreen")

    # Display player names
    game_title = tk.Label(
        game_window,
        text="GAME STARTED",
        font=("Helvetica", 20, "bold"),
        bg="lightgreen"
    )
    game_title.pack(pady=20)

    players_label = tk.Label(
        game_window,
        text=f"{player1_name}  VS  {player2_name}",
        font=("Helvetica", 16, "bold"),
        bg="lightgreen"
    )
    players_label.pack(pady=20)


# ==========================================
# Function: Open Player Information Window
# ==========================================
def player_info():
    """
    Opens a window where users can
    enter player names.
    """

    # Create a new window
    player_window = tk.Toplevel()

    # Configure window
    player_window.title("Player Information")
    player_window.geometry("500x300")
    player_window.configure(bg="lightblue")

    # Variables to store player names
    player1_name_var = tk.StringVar()
    player2_name_var = tk.StringVar()

    # --------------------------------------
    # Heading
    # --------------------------------------
    heading_label = tk.Label(
        player_window,
        text="Enter Player Names",
        font=("Helvetica", 18, "bold"),
        bg="lightblue"
    )
    heading_label.grid(row=0, column=0, columnspan=2, pady=20)

    # --------------------------------------
    # Player 1 Label
    # --------------------------------------
    player1_label = tk.Label(
        player_window,
        text="Player 1",
        font=("Helvetica", 12, "bold"),
        bg="black",
        fg="white",
        width=12
    )
    player1_label.grid(row=1, column=0, padx=20, pady=10)

    # --------------------------------------
    # Player 1 Entry Box
    # --------------------------------------
    player1_entry = tk.Entry(
        player_window,
        textvariable=player1_name_var,
        width=20
    )
    player1_entry.grid(row=2, column=0, padx=20)

    # --------------------------------------
    # Player 2 Label
    # --------------------------------------
    player2_label = tk.Label(
        player_window,
        text="Player 2",
        font=("Helvetica", 12, "bold"),
        bg="black",
        fg="white",
        width=12
    )
    player2_label.grid(row=1, column=1, padx=20, pady=10)

    # --------------------------------------
    # Player 2 Entry Box
    # --------------------------------------
    player2_entry = tk.Entry(
        player_window,
        textvariable=player2_name_var,
        width=20
    )
    player2_entry.grid(row=2, column=1, padx=20)

    # --------------------------------------
    # Submit Button Function
    # --------------------------------------
    def submit_names():
        """
        Gets player names and opens
        the game screen.
        """

        player1_name = player1_name_var.get()
        player2_name = player2_name_var.get()

        print("Player 1:", player1_name)
        print("Player 2:", player2_name)

        # Close player info window
        player_window.destroy()

        # Open game screen
        open_game_screen(player1_name, player2_name)

    # --------------------------------------
    # Submit Button
    # --------------------------------------
    submit_button = tk.Button(
        player_window,
        text="Submit",
        command=submit_names,
        bg="green",
        fg="white",
        font=("Helvetica", 12, "bold")
    )
    submit_button.grid(
        row=3,
        column=0,
        columnspan=2,
        pady=30
    )


# ==========================================
# Main Application Window
# ==========================================

# Create root window
main_window = tk.Tk()

# Window title
main_window.title("My Game")

# Window size
main_window.geometry("500x550")

# Background color
main_window.configure(bg="lightblue")


# ==========================================
# Game Image Canvas
# ==========================================
game_canvas = tk.Canvas(
    main_window,
    width=300,
    height=300,
    bg="white"
)
game_canvas.pack(pady=20)


# ==========================================
# Load and Display Image
# ==========================================
try:
    player_image = tk.PhotoImage(
        master=main_window,
        file="/Users/ravirajpanchal/Learn Pygame/Run_Left/tile000.png"
    )

    game_canvas.create_image(
        150,
        150,
        image=player_image
    )

except Exception as error:
    print("Image could not be loaded.")
    print(error)


# ==========================================
# Game Title
# ==========================================
game_title_label = tk.Label(
    main_window,
    text="GAME NAME",
    font=("Helvetica", 24, "bold"),
    fg="red",
    bg="yellow"
)
game_title_label.place(x=150, y=10)


# ==========================================
# Button Frame
# ==========================================
button_frame = tk.Frame(
    main_window,
    bg="lightblue"
)
button_frame.pack(side=tk.BOTTOM, pady=20)


# ==========================================
# Start Game Button
# ==========================================
start_button = tk.Button(
    button_frame,
    text="Start Game",
    command=player_info,
    bg="lightgreen",
    fg="black",
    font=("Helvetica", 12, "bold")
)
start_button.pack(padx=10, pady=10)


# ==========================================
# Run Application
# ==========================================
main_window.mainloop()
```
