```python
# Import required modules
import pygame
import os
import sys
import math
from pygame.locals import *

# Initialize all pygame modules
pygame.init()

# -------------------------------------------------------------------
# WINDOW SETUP
# -------------------------------------------------------------------

# Create the game window size
SIZE = (1000, 640)

# Create the display window
display = pygame.display.set_mode(SIZE)

# Set the window title
pygame.display.set_caption("Image Transformation")

# Create a clock object to control frame rate
clock = pygame.time.Clock()

# Define colors
PINK = (255, 123, 255)
BLACK = (0, 0, 0)

# Fill the background with black once
display.fill(BLACK)

# -------------------------------------------------------------------
# IMAGE LOADING
# -------------------------------------------------------------------

# Load the image
picture = pygame.image.load('/users/ravirajpanchal/Learn Pygame/PianoKeysFunction.gif').convert_alpha()

# Draw a pink border around the image
rect0 = picture.get_rect()
pygame.draw.rect(picture, PINK, rect0, 5)

# Get display center
w, h = SIZE
center = (w // 2, h // 2)

# Create image and rectangle objects
image = picture
rect = image.get_rect()
rect.center = center

# -------------------------------------------------------------------
# GAME VARIABLES
# -------------------------------------------------------------------

# Variable used to control the game loop
stop = False

# Rotation angle
angle = 0

# Scale factor
scale = 1

# -------------------------------------------------------------------
# MAIN GAME LOOP
# -------------------------------------------------------------------

while not stop:

    # Get all events from the game window
    for event in pygame.event.get():

        # Check if the user clicks the close button
        if event.type == pygame.QUIT:
            stop = True

        # Check keyboard input
        if event.type == pygame.KEYDOWN:

            # -------------------------------------------------------
            # ROTATE IMAGE (R KEY)
            # SHIFT + R = rotate left
            # R = rotate right
            # -------------------------------------------------------
            if event.key == K_r:

                if event.mod & KMOD_SHIFT:
                    angle -= 20
                else:
                    angle += 20

                image = pygame.transform.rotozoom(picture, angle, scale)

            # -------------------------------------------------------
            # SCALE IMAGE (T KEY)
            # SHIFT + T = zoom out
            # T = zoom in
            # -------------------------------------------------------
            elif event.key == K_t:

                if event.mod & KMOD_SHIFT:
                    scale /= 1.1
                else:
                    scale *= 1.1

                image = pygame.transform.rotozoom(picture, angle, scale)

            # -------------------------------------------------------
            # RESET IMAGE (O KEY)
            # -------------------------------------------------------
            elif event.key == K_o:

                image = picture
                scale = 1
                angle = 0

            # -------------------------------------------------------
            # FLIP IMAGE (0 KEY)
            # -------------------------------------------------------
            elif event.key == K_h:

                # Flip horizontally
                image = pygame.transform.flip(image, True, False)

            elif event.key == K_v:
                
                # Flip vertically
                image = pygame.transform.flip(image, False, True)

            # -------------------------------------------------------
            # APPLY LAPLACIAN FILTER (L KEY)
            # -------------------------------------------------------
            elif event.key == K_l:

                image = pygame.transform.laplacian(image)

            # -------------------------------------------------------
            # SCALE IMAGE 2X (S KEY)
            # -------------------------------------------------------
            elif event.key == K_s:

                image = pygame.transform.scale2x(image)

            # -------------------------------------------------------
            # ROTATE IMAGE 90 DEGREES (D KEY)
            # -------------------------------------------------------
            elif event.key == K_d:

                image = pygame.transform.rotate(image, 90)

            # Update rectangle after transformations
            rect = image.get_rect()
            rect.center = center

    # ----------------------------------------------------------------
    # DRAW EVERYTHING
    # ----------------------------------------------------------------

    # Clear screen
    display.fill(BLACK)

    # Draw image
    display.blit(image, rect)

    # Update the game display screen
    pygame.display.update()

    # Set the frame rate to 40 FPS
    clock.tick(40)

# Quit pygame properly
pygame.quit()


```
