```python
# Import OpenGL functions
from OpenGL.GL import *
from OpenGL.GLU import *  # Needed for gluPerspective

# Import pygame
import pygame
from pygame.locals import *

# Vertices of a triangular pyramid (tetrahedron-like shape)
vertices = (
    (1, -1, -1),    # Vertex 0
    (1, -1, 1),     # Vertex 1
    (-1, -1, 1),    # Vertex 2
    (-1, 1, 1),     # Vertex 3
    (-1, -1, -1)    # Vertex 4 (apex)
)

# Edges connecting the vertices
edges = (
    (0, 1),
    (1, 2),
    (2, 3),
    (3, 0),
    (0, 4),
    (1, 4),
    (2, 4),
    (3, 4)
)

def Triangle():
    """
    Draws the wireframe shape using line segments.
    """
    
    # Start drawing lines
    glBegin(GL_LINES)

    # Draw each edge
    for edge in edges:
        for vertex in edge:

            # Send vertex coordinates to OpenGL
            glVertex3fv(vertices[vertex])

    # Finish drawing
    glEnd()


def main():
    """
    Main game/rendering loop.
    """

    # Initialize pygame
    pygame.init()

    # Window size
    screen_size = (700, 600)

    # Create OpenGL display window
    pygame.display.set_mode(screen_size, DOUBLEBUF | OPENGL)

    # Set perspective projection
    gluPerspective(
        45,                     # Field of view
        screen_size[0] / screen_size[1],  # Aspect ratio
        0.1,                    # Near clipping plane
        50.0                    # Far clipping plane
    )

    # Move object away from camera
    glTranslatef(0.0, 0.0, -5)

    # Main rendering loop
    running = True

    while running:

        # Process events
        for event in pygame.event.get():

            # Close window
            if event.type == pygame.QUIT:
                running = False

        # Rotate object slightly each frame
        glRotatef(1, 3, 1, 1)

        # Clear screen and depth buffer
        glClear(GL_COLOR_BUFFER_BIT | GL_DEPTH_BUFFER_BIT)

        # Draw the shape
        Triangle()

        # Update display
        pygame.display.flip()

        # Limit speed (~100 FPS)
        pygame.time.wait(10)

    # Clean shutdown
    pygame.quit()


# Start program
main()
```
