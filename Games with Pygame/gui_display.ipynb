```python
# ==========================================
# Simple Game Menu using Tkinter
# ==========================================

# Import the Tkinter library
import tkinter as tk
#import funtion player_info from gui_display_w11.py
from gui_display_w11 import player_info

# ------------------------------------------
# Create the main application window
# ------------------------------------------
main_window = tk.Tk()

# Set the title displayed at the top of the window
main_window.title("My Game")

# Set the size and screen position of the window
# Format: Width x Height + X_Position + Y_Position
main_window.geometry("400x500+100+100")

# Set the background color of the window
main_window.configure(bg="lightblue")


# ------------------------------------------
# Create a canvas to display graphics/images
# ------------------------------------------
game_canvas = tk.Canvas(
    main_window,
    width=300,
    height=350,
    bg="white"
)

game_canvas.pack(pady=10)


# ------------------------------------------
# Load and display an image
# ------------------------------------------
# The master=main_window parameter helps
# prevent image-related errors in Jupyter.
# ------------------------------------------
try:
    player_image = tk.PhotoImage(
        master=main_window,
        file="/Users/ravirajpanchal/Learn Pygame/Run_Left/tile000.png"
    )

    # Display the image in the center of the canvas
    game_canvas.create_image(
        150,      # x-coordinate
        175,      # y-coordinate
        image=player_image
    )

except Exception as error:
    print("Unable to load image:")
    print(error)


# ------------------------------------------
# Create a title label
# ------------------------------------------
game_title_label = tk.Label(
    main_window,
    text="New Game",
    font=("Helvetica", 24, "bold"),
    fg="red",
    bg="yellow"
)

game_title_label.place(x=100, y=10)


# ------------------------------------------
# Create a frame to hold buttons
# ------------------------------------------
button_frame = tk.Frame(
    main_window,
    bg="lightblue"
)

button_frame.pack(side=tk.BOTTOM, pady=20)


# ------------------------------------------
# Create the Start Game button
# ------------------------------------------
start_button = tk.Button(
    button_frame,
    text="Start Game",
    command=player_info, # fucntion to call for player info
    bg="lightgreen",
    fg="black",
    font=("Helvetica", 12)
)

start_button.pack(padx=10, pady=10)


# ------------------------------------------
# Start the Tkinter event loop
# ------------------------------------------
main_window.mainloop()
```
