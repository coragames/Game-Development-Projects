```python
# Import the pygame module so we can create games and graphics
import pygame

# Initialize all imported pygame modules
pygame.init()

# -------------------------------------------------------------------
# WINDOW SETUP
# -------------------------------------------------------------------

# Define the width and height of the game window
SIZE = (1000, 640)

# Create the game display window using the SIZE dimensions
display = pygame.display.set_mode(SIZE)

# Set the title that appears at the top of the game window
pygame.display.set_caption("Fonts")

# Create a clock object to help control the frame rate (speed of the loop)
clock = pygame.time.Clock()

# Boolean variable used to keep the game loop running
running = True

# Create a font object
# "comicsans" = font style
# 70 = font size
font = pygame.font.SysFont("comicsans", 70)

# Render the text into an image surface
# "GAME" = text displayed
# True = smooth/anti-aliased text
# (0, 128, 0) = green text color in RGB
text = font.render("GAME", True, (0, 128, 0))

# Get a list of all available system fonts
# font_types = pygame.font.get_fonts()
# print(font_types)

# Get the default pygame font name
# font_name = pygame.font.get_default_font()
# print(font_name)

# Main game loop
while running:

    # Get all events currently happening in the game window
    for event in pygame.event.get():

        # Check if the user clicks the window close button
        if event.type == pygame.QUIT:
            running = False

        # Check if a key is pressed and if it is the ESCAPE key
        # If ESCAPE is pressed, stop the game loop
        if event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE:
            running = False

    # Fill the screen with a yellow background color
    display.fill((255, 255, 0))    

    # Draw the rendered text onto the display window
    # (100, 100) = x and y position on the screen
    display.blit(text, (100, 100))

    # Update the entire display so changes appear on screen
    pygame.display.update()

    # Limit the loop to 40 frames per second
    clock.tick(40)

# Quit pygame and close the window properly
pygame.quit()
```
