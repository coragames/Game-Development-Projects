```python
import pygame
import sys

# -----------------------------
# Initialize pygame modules
# -----------------------------
pygame.init()

# -----------------------------
# Window settings
# -----------------------------
WIDTH = 1000
HEIGHT = 1000
SIZE = (WIDTH, HEIGHT)

# Create the game window
display = pygame.display.set_mode(SIZE)

# Set the window title
pygame.display.set_caption("Animation")

# -----------------------------
# Load image
# -----------------------------
# convert_alpha() keeps transparency
pic = pygame.image.load("MPSS_Mario.jpg").convert_alpha()

# -----------------------------
# Clock for controlling FPS
# -----------------------------
clock = pygame.time.Clock()

# -----------------------------
# Game loop control variable
# -----------------------------
running = True

# -----------------------------
# Starting position of image
# Center the image on screen
# -----------------------------
pos_x = (WIDTH - pic.get_width()) // 2
pos_y = (HEIGHT - pic.get_height()) // 2

# Movement speed
speed = 5

# =============================
# Main Game Loop
# =============================
while running:

    # Fill the screen with white
    display.fill((255, 255, 255))

    # -------------------------
    # Event handling
    # -------------------------
    for event in pygame.event.get():

        # Close window button pressed
        if event.type == pygame.QUIT:
            running = False

        # ESC key pressed
        if event.type == pygame.KEYDOWN:
            if event.key == pygame.K_ESCAPE:
                running = False

    # -------------------------
    # Continuous key movement
    # -------------------------
    keys = pygame.key.get_pressed()

    # Move left
    if keys[pygame.K_a]:
        pos_x -= speed

    # Move right
    if keys[pygame.K_d]:
        pos_x += speed

    # Move up
    if keys[pygame.K_w]:
        pos_y -= speed

    # Move down
    if keys[pygame.K_s]:
        pos_y += speed

    # -------------------------
    # Keep image inside window
    # -------------------------
    pos_x = max(0, min(pos_x, WIDTH - pic.get_width()))
    pos_y = max(0, min(pos_y, HEIGHT - pic.get_height()))

    # -------------------------
    # Draw image on screen
    # -------------------------
    display.blit(pic, (pos_x, pos_y))

    # Update the display
    pygame.display.update()

    # Limit frame rate to 60 FPS
    clock.tick(60)

# -----------------------------
# Quit pygame properly
# -----------------------------
pygame.quit()

# Exit the program
sys.exit()
```
