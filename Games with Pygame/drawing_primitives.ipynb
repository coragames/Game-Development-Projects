```python
# Import the pygame library
import pygame

# Initialize all pygame modules
pygame.init()

# Create the screen size using width and height
screen_size = [400, 600]

# Create the display window
screen_display = pygame.display.set_mode(screen_size)

# Set the title of the game window
pygame.display.set_caption("Primitive Shapes")

# Variable used to stop the game loop
stop = False

# Create a clock object to control game speed
clock = pygame.time.Clock()

# Main game loop
while not stop:

    # Check all events happening in the game window
    for event in pygame.event.get():

        # If the user clicks the close button
        if event.type == pygame.QUIT:
            stop = True

    # Set the frame rate to 60 FPS
    clock.tick(60)

    # Fill the screen with blue background color
    screen_display.fill((0, 0, 255))

    # Draw a white rectangle
    # Format: (surface, color, [x-position, y-position, width, height])
    pygame.draw.rect(screen_display, (255, 255, 255), [85, 40, 30, 10])

    # Update the display to show changes
    pygame.display.update()

# Quit pygame properly after the loop ends
pygame.quit()
```
