```python
# Import the pygame library
import pygame

# Initialize all pygame modules
pygame.init()

# Create the game window size
size = (600, 600)

# Create the display window
display = pygame.display.set_mode(size)

# Set the window title
pygame.display.set_caption("Draw with Mouse")

# Create a clock object to control frame rate
clock = pygame.time.Clock()

# Fill the background with black color once
display.fill((0, 0, 0))

# Variable to control the game loop
stop = False

# Main game loop
while not stop:
    try:
        # Check all events happening in the window
        for event in pygame.event.get():

            # If user clicks the close button
            if event.type == pygame.QUIT:
                stop = True

            # If mouse button is pressed
            if event.type == pygame.MOUSEBUTTONDOWN:

                # Get mouse position
                px, py = pygame.mouse.get_pos()

                # Draw a green circle where mouse is clicked
                pygame.draw.circle(display, (0, 255, 0), (px, py), 20)

            # If mouse button is released
            if event.type == pygame.MOUSEBUTTONUP:
                print("Mouse button released")

        # Update the display screen
        pygame.display.update()

        # Limit the loop to 60 frames per second
        clock.tick(60)

    # Handle any unexpected errors
    except Exception as e:
        print(e)

# Quit pygame properly
pygame.quit()



    
        
```
