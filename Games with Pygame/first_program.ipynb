```python
# Import the pygame module
import pygame

# Initialize all pygame modules
pygame.init()

# Create the game display window
# (400 = width, 800 = height)
game_display = pygame.display.set_mode((400, 800))

# Set the title of the game window
pygame.display.set_caption("my first game app")

# Create a clock object to control frame rate
clock = pygame.time.Clock()

# Variable used to control the game loop
stop = False

# Main game loop
while not stop:

    # Get all events from the game window
    for event in pygame.event.get():

        # Print every event in the console
        print(event)

        # Check if the user clicks the close button
        if event.type == pygame.QUIT:
            stop = True

    # Update the game display screen
    pygame.display.update()

    # Set the frame rate to 40 FPS
    clock.tick(40)

# Quit pygame properly
pygame.quit()
```
