```python
import pygame
import sys

# -----------------------------
# Initialize pygame modules
# -----------------------------
pygame.init()

# Screen size (width x height)
WIDTH = 1000
HEIGHT = 1000

# Create game window
display = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Moving Character")

# Clock helps control game speed (FPS)
clock = pygame.time.Clock()

# =====================================================
# LOAD SPRITES (IMAGES FOR ANIMATION)
# =====================================================

# Left movement animation frames
left_img = [
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile000.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile001.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile002.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile003.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile004.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile005.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile006.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile007.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile008.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile009.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile010.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile011.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile012.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile013.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile014.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Right/tile015.png")
]

# Right movement animation frames
right_img = [
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile000.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile001.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile002.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile003.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile004.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile005.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile006.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile007.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile008.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile009.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile010.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile011.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile012.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile013.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile014.png"),
    pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile015.png")
]

# Idle image (standing still)
idle_img = pygame.image.load("/Users/ravirajpanchal/Learn Pygame/Run_Left/tile000.png")

# Play BG Music continously
bg_music = pygame.mixer.music.load("/Users/ravirajpanchal/Learn Pygame/(Loop) juanjo_sound - matumbia.wav")
pygame.mixer.music.play(-1)

# Jump SFX
jump_sfx = pygame.mixer.Sound("/Users/ravirajpanchal/Learn Pygame/Interface Push Button.mp3")

# =====================================================
# CHARACTER SETTINGS
# =====================================================

# Starting position (center of screen)
pos_x = WIDTH // 2
pos_y = HEIGHT // 2

# Ground position
GROUND_Y = HEIGHT // 2

# Movement speed (how fast character moves)
speed = 5

# Animation frame tracker
frame = 0

# Controls how fast animation changes frames
animation_speed = 0.3

# Jump variables
jump_velocity = 0
gravity = 0.8
is_jumping = False

# =====================================================
# FUNCTION: MOVE CHARACTER
# =====================================================

def move_character(keys, x, y):
    global frame
    global jump_velocity
    global is_jumping

    moving_left = False
    moving_right = False

    # Left movement
    if keys[pygame.K_LEFT] or keys[pygame.K_a]:
        x -= speed
        moving_left = True

    # Right movement
    elif keys[pygame.K_RIGHT] or keys[pygame.K_d]:
        x += speed
        moving_right = True

    # Start jump
    if keys[pygame.K_SPACE] and not is_jumping:
        jump_velocity = -15
        is_jumping = True
        jump_sfx.play()

    # Apply gravity
    jump_velocity += gravity
    y += jump_velocity

    # Land on ground
    if y >= GROUND_Y:
        y = GROUND_Y
        jump_velocity = 0
        is_jumping = False

    # Keep inside screen
    x = max(0, min(x, WIDTH - 100))

    # Animation
    if moving_left or moving_right:
        frame += animation_speed

    if moving_right:
        if frame >= len(right_img):
            frame = 0
        img = right_img[int(frame)]

    elif moving_left:
        if frame >= len(left_img):
            frame = 0
        img = left_img[int(frame)]

    else:
        img = idle_img

    return x, y, img

# =====================================================
# MAIN GAME LOOP
# =====================================================

running = True
current_img = idle_img

while running:

    # Fill screen with white each frame
    display.fill((255, 125, 255))

    # Check events (like closing window)
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Get keyboard input
    keys = pygame.key.get_pressed()

    # Move character and get updated image
    pos_x, pos_y, current_img = move_character(keys, pos_x, pos_y)

    # Draw character on screen
    display.blit(current_img, (pos_x, pos_y))

    # Refresh screen
    pygame.display.update()

    # Limit FPS to 60 frames per second
    clock.tick(60)

# Exit the program
pygame.quit()
```
