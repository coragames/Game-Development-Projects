```python
import pygame
import random
import sys

# ---------------------------
# Initialize pygame
# ---------------------------
pygame.init()

# ---------------------------
# Game settings
# ---------------------------
SCREEN_WIDTH = 600
SCREEN_HEIGHT = 400

GAME_SCREEN = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))
pygame.display.set_caption("Snake Game")

# -------------------------------
# Setup background music and sfx
# -------------------------------
pygame.mixer.music.load("/Users/ravirajpanchal/Learn Pygame/(Loop) juanjo_sound - matumbia.wav")
pygame.mixer.music.play(-1)

food_sfx = pygame.mixer.Sound("/Users/ravirajpanchal/Learn Pygame/Interface Push Button.mp3")

# ---------------------------
# Colors
# ---------------------------
BLACK = (0, 0, 0)
GREEN = (0, 255, 0)
RED = (255, 0, 0)
WHITE = (255, 255, 255)

# ---------------------------
# Snake settings
# ---------------------------
BLOCK_SIZE = 20

# Game speed (level)
game_level = int(input("Enter game level (5-30): "))

# ---------------------------
# Fonts
# ---------------------------
FONT = pygame.font.SysFont("calibri", 25)

# ---------------------------
# Clock
# ---------------------------
clock = pygame.time.Clock()


def draw_text(message, color, x, y):
    """
    Display text on screen
    """
    text_surface = FONT.render(message, True, color)
    GAME_SCREEN.blit(text_surface, (x, y))


def display_score(score):
    """
    Display current score
    """
    score_text = FONT.render(f"Score: {score}", True, RED)
    GAME_SCREEN.blit(score_text, (10, 10))


def draw_snake(snake_blocks):
    """
    Draw snake body
    """
    for block in snake_blocks:
        pygame.draw.rect(
            GAME_SCREEN,
            GREEN,
            [block[0], block[1], BLOCK_SIZE, BLOCK_SIZE]
        )

def game_loop():

    game_over = False
    game_close = False

    # Snake starting position
    snake_x = SCREEN_WIDTH // 2
    snake_y = SCREEN_HEIGHT // 2

    # Movement variables
    x_change = 0
    y_change = 0 

    # Snake body
    snake_body = []
    snake_length = 1

    # Food position
    food_x = round(
        random.randrange(0, SCREEN_WIDTH - BLOCK_SIZE)
        / BLOCK_SIZE
    ) * BLOCK_SIZE

    food_y = round(
        random.randrange(0, SCREEN_HEIGHT - BLOCK_SIZE)
        / BLOCK_SIZE
    ) * BLOCK_SIZE

    # ---------------------------
    # Main Game Loop
    # ---------------------------
    while not game_over:

        # -----------------------
        # Game Over Screen
        # -----------------------
        while game_close:

            GAME_SCREEN.fill(BLACK)

            draw_text(
                "Game Over!",
                RED,
                SCREEN_WIDTH // 2 - 70,
                SCREEN_HEIGHT // 3
            )

            draw_text(
                "Press C to Play Again",
                WHITE,
                170,
                SCREEN_HEIGHT // 2
            )

            draw_text(
                "Press Q to Quit",
                WHITE,
                210,
                SCREEN_HEIGHT // 2 + 40
            )

            display_score(snake_length - 1)

            pygame.display.update()

            for event in pygame.event.get():

                if event.type == pygame.QUIT:
                    pygame.quit()
                    sys.exit()

                if event.type == pygame.KEYDOWN:

                    if event.key == pygame.K_q:
                        pygame.quit()
                        sys.exit()

                    if event.key == pygame.K_c:
                        game_loop()

        # -----------------------
        # Event Handling
        # -----------------------
        for event in pygame.event.get():

            if event.type == pygame.QUIT:
                game_over = True

            if event.type == pygame.KEYDOWN:

                # Move left
                if event.key == pygame.K_LEFT:
                    x_change = -BLOCK_SIZE
                    y_change = 0

                # Move right
                elif event.key == pygame.K_RIGHT:
                    x_change = BLOCK_SIZE
                    y_change = 0

                # Move up
                elif event.key == pygame.K_UP:
                    x_change = 0
                    y_change = -BLOCK_SIZE

                # Move down
                elif event.key == pygame.K_DOWN:
                    x_change = 0
                    y_change = BLOCK_SIZE

                # Pause game
                elif event.key == pygame.K_p:
                    pause_game()

                # Quit game
                elif event.key == pygame.K_q:
                    game_over = True

        # -----------------------
        # Update snake position
        # -----------------------
        snake_x += x_change
        snake_y += y_change

        # -----------------------
        # Wall collision/wrapping
        # -----------------------
        # Left edge -> Right edge
        if snake_x < 0:
            snake_x = SCREEN_WIDTH - BLOCK_SIZE
        
        # Right edge -> Left edge
        elif snake_x >= SCREEN_WIDTH:
            snake_x = 0
        
        # Top edge -> Bottom edge
        if snake_y < 0:
            snake_y = SCREEN_HEIGHT - BLOCK_SIZE
        
        # Bottom edge -> Top edge
        elif snake_y >= SCREEN_HEIGHT:
            snake_y = 0

        # -----------------------
        # Draw screen
        # -----------------------
        GAME_SCREEN.fill(BLACK)

        # Draw food
        pygame.draw.circle(
        GAME_SCREEN,
        RED,
        (food_x + BLOCK_SIZE // 2, food_y + BLOCK_SIZE // 2),
        BLOCK_SIZE // 2
        )

        # Snake head
        snake_head = []
        snake_head.append(snake_x)
        snake_head.append(snake_y)

        snake_body.append(snake_head)

        # Keep correct length
        if len(snake_body) > snake_length:
            del snake_body[0]

        # Self collision
        for block in snake_body[:-1]:
            if block == snake_head:
                game_close = True

        # Draw snake
        draw_snake(snake_body)

        # Draw score
        display_score(snake_length - 1)

        pygame.display.update()

        # -----------------------
        # Food collision
        # -----------------------
        if snake_x == food_x and snake_y == food_y:
            
            food_x = round(
                random.randrange(
                    0,
                    SCREEN_WIDTH - BLOCK_SIZE
                ) / BLOCK_SIZE
            ) * BLOCK_SIZE

            food_y = round(
                random.randrange(
                    0,
                    SCREEN_HEIGHT - BLOCK_SIZE
                ) / BLOCK_SIZE
            ) * BLOCK_SIZE

            snake_length += 1
            food_sfx.play()
            
        # Control speed
        clock.tick(game_level)

    pygame.quit()
    sys.exit()


# ---------------------------
# Start game
# ---------------------------
game_loop()
```
